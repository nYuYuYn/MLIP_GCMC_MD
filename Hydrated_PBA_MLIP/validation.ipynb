```python
# Separate the cfg file into section
# Load Each section one by one
def load_cfg(filename):
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()[4:-5]  # Read from the 5th line onwards 
        
        # Split each line into separate numbers
        numbers = []
        for line in lines:
            line_numbers = [float(num) for num in line.split()]  # Assuming numbers are floats
            numbers.append(line_numbers)
        
        # Convert the list of lists into a NumPy array
        return np.array(numbers)
    
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except IsADirectoryError:
        print(f"Error: '{filename}' is a directory, not a file.")
    except IOError:
        print(f"Error: Could not read file '{filename}'.")
    except Exception as e:
        print(f"Error: {e}")
```

```python
def calculate_rmse(actual, predicted):
    mse = np.mean((actual - predicted) ** 2)
    rmse = np.sqrt(mse)
    return rmse
```

```python
def flatten(lst):
    flattened = []
    for item in lst:
        if isinstance(item, list):
            flattened.extend(flatten(item))
        else:
            flattened.append(item)
    return flattened
```

```python
def convert_dump_file(input_file, output_file, element_map):
    """
    Convert atom types to element names in a LAMMPS dump file.
    
    Parameters:
    input_file (str): Path to the original LAMMPS dump file
    output_file (str): Path to save the modified file
    element_map (dict): Dictionary mapping atom types to element names
    """
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        in_atoms_section = False
        
        for line in infile:
            line = line.strip()
            
            # Check if we're starting the atoms section
            if line.startswith("ITEM: ATOMS"):
                in_atoms_section = True
                outfile.write(line + '\n')
                continue
            
            # Process atom lines when we're in the atoms section
            if in_atoms_section and not line.startswith("ITEM:"):
                parts = line.split()
                
                if len(parts) >= 2:  # Ensure we have at least atom ID and type
                    # Get the atom type (2nd column in your file)
                    atom_type = int(parts[1])
                    
                    # Replace type number with element name if in our map
                    if atom_type in element_map:
                        parts[1] = element_map[atom_type]
                    
                    # Write the modified line
                    outfile.write(' '.join(parts) + '\n')
                else:
                    # Just write the line unchanged if it doesn't match expected format
                    outfile.write(line + '\n')
            else:
                # If we encounter a new "ITEM:" line, we're leaving the atoms section
                if line.startswith("ITEM:") and not line.startswith("ITEM: ATOMS"):
                    in_atoms_section = False
                
                # Write all non-atom lines unchanged
                outfile.write(line + '\n')
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

mtp_file = './validation/cal_hydration_16.cfg.0'
dft_file = './validation/hydration/hydration.cfg'

E_MTP, _, F_MTP, _ = load_multi_section_cfg_no_grade(mtp_file)
E_DFT, _, F_DFT, _ = load_multi_section_cfg_no_grade(dft_file)

E_MTP = np.asarray(E_MTP, dtype=float)
E_DFT = np.asarray(E_DFT, dtype=float)

if len(F_DFT) != len(F_MTP) or len(E_DFT) != len(E_MTP):
    raise ValueError('The DFT and MTP files have different numbers of configurations.')

atom_counts_dft = np.array([np.asarray(f).shape[0] for f in F_DFT])
atom_counts_mtp = np.array([np.asarray(f).shape[0] for f in F_MTP])

if not np.array_equal(atom_counts_dft, atom_counts_mtp):
    raise ValueError('The DFT and MTP atom counts do not match.')

# All x, y, and z force components, across all atoms and configurations
FDFT_components = np.concatenate([
    np.asarray(f, dtype=float)[:, 5:].ravel() for f in F_DFT
])
FMTP_components = np.concatenate([
    np.asarray(f, dtype=float)[:, 5:].ravel() for f in F_MTP
])

# Energy per atom
E_DFT_per_atom = E_DFT / atom_counts_dft
E_MTP_per_atom = E_MTP / atom_counts_dft

# Errors and metrics
force_errors = FMTP_components - FDFT_components
energy_errors = E_MTP_per_atom - E_DFT_per_atom

force_mae = np.mean(np.abs(force_errors))
force_rmse = np.sqrt(np.mean(force_errors**2))
force_r2 = stats.linregress(
    FDFT_components, FMTP_components
).rvalue**2

energy_mae = np.mean(np.abs(energy_errors))
energy_rmse = np.sqrt(np.mean(energy_errors**2))
energy_r2 = stats.linregress(
    E_DFT_per_atom, E_MTP_per_atom
).rvalue**2
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Function to read and process each file
def read_battery_data(filename):
    """Read battery data file and return cleaned DataFrame"""
    try:
        # Read the file, skipping the header row
        df = pd.read_csv(filename, sep='\t', skiprows=0)
        
        # Clean column names (remove extra spaces)
        df.columns = df.columns.str.strip()
        
        # Convert Publication Years to numeric, handling any non-numeric values
        df['Publication Years'] = pd.to_numeric(df['Publication Years'], errors='coerce')
        df['Record Count'] = pd.to_numeric(df['Record Count'], errors='coerce')
        
        # Remove rows with NaN values
        df = df.dropna()
        
        # Sort by year for better visualization
        df = df.sort_values('Publication Years')
        
        return df
    except Exception as e:
        print(f"Error reading {filename}: {e}")
        return None

# Read all three files
files = {
    'MD': 'MD_battery.txt',
    'AIMD': 'AIMD_battery.txt', 
    'MLIP': 'MLIP_battery.txt'
}

data = {}
for key, filename in files.items():
    df = read_battery_data(filename)
    if df is not None:
        data[key] = df
        print(f"Successfully loaded {filename}: {len(df)} records")
    else:
        print(f"Failed to load {filename}")

# Create visualization if data was loaded successfully
if data:
    # Set up the plot
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(15, 12))
    
    # Plot 1: Separate bar charts for each method
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c']  # Blue, Orange, Green
    
    for i, (method, df) in enumerate(data.items()):
        # Filter to years with significant data (e.g., > 5 publications)
        df_filtered = df[df['Record Count'] >= 5].copy()
        total_pubs = df['Record Count'].sum()
        
        ax1.bar(df_filtered['Publication Years'] + i*0.25, df_filtered['Record Count'], 
                width=0.25, label=f'{method} Battery Research (n={total_pubs})', color=colors[i], alpha=0.8)
    
    ax1.set_xlabel('Publication Year')
    ax1.set_ylabel('Number of Publications')
    ax1.set_title('Battery Research Publications by Method Over Time')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    
    # Plot 2: Line plot to show trends more clearly
    for i, (method, df) in enumerate(data.items()):
        df_filtered = df[df['Record Count'] >= 5].copy()
        total_pubs = df['Record Count'].sum()
        
        ax2.plot(df_filtered['Publication Years'], df_filtered['Record Count'], 
                marker='o', linewidth=2, label=f'{method} Battery Research (n={total_pubs})', color=colors[i])
    
    ax2.set_xlabel('Publication Year')
    ax2.set_ylabel('Number of Publications')
    ax2.set_title('Publication Trends in Battery Research Methods')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Print summary statistics
    print("\n" + "="*50)
    print("SUMMARY STATISTICS")
    print("="*50)
    
    for method, df in data.items():
        total_pubs = df['Record Count'].sum()
        peak_year = df.loc[df['Record Count'].idxmax(), 'Publication Years']
        peak_count = df['Record Count'].max()
        
        print(f"\n{method} Battery Research:")
        print(f"  Total Publications: {total_pubs}")
        print(f"  Peak Year: {int(peak_year)} ({peak_count} publications)")
        print(f"  Years with Data: {int(df['Publication Years'].min())} - {int(df['Publication Years'].max())}")
    
    # Create a combined comparison for recent years (2015-2025)
    print(f"\n{'Method':<8} {'2020':<8} {'2021':<8} {'2022':<8} {'2023':<8} {'2024':<8} {'2025':<8}")
    print("-" * 60)
    
    for method, df in data.items():
        row = f"{method:<8}"
        for year in [2020, 2021, 2022, 2023, 2024, 2025]:
            count = df[df['Publication Years'] == year]['Record Count'].values
            count_str = str(int(count[0])) if len(count) > 0 else "0"
            row += f"{count_str:<8}"
        print(row)

else:
    print("No data files could be loaded. Please check file names and format.")
```

```python
import argparse
import sys
import re

def parse_args():
    p = argparse.ArgumentParser(
        description="Convert a single‐frame LAMMPS data file (atom_style atomic) "
                    "into one MLIP‐MTP .cfg file."
    )
    p.add_argument(
        "--in", "-i", dest="infile", required=True,
        help="Path to input LAMMPS data file (atom_style atomic)."
    )
    p.add_argument(
        "--out", "-o", dest="outfile", required=True,
        help="Path to output MLIP .cfg file."
    )
    p.add_argument(
        "--map", "-m", dest="type_map", required=True,
        help="Comma‐separated mapping of LAMMPS type→element, e.g. 1=Fe,2=Mn,3=C"
    )
    return p.parse_args()

def read_type_map(s):
    """
    Parse a string like "1=Fe,2=Mn,3=C,4=N,5=Na" into a dict {1:"Fe",2:"Mn",...}.
    """
    mp = {}
    for pair in s.strip().split(","):
        if '=' not in pair:
            continue
        t, sym = pair.split("=")
        try:
            t_int = int(t)
        except ValueError:
            raise ValueError(f"Cannot parse type ID '{t}' as integer in mapping '{s}'")
        mp[t_int] = sym
    return mp

def parse_lammps_data(filename):
    """
    Very minimal LAMMPS-data parser (atom_style atomic).
    Returns:
      • natoms        : int
      • box_bounds    : ((xlo,xhi),(ylo,yhi),(zlo,zhi))
      • atoms         : list of (id, type, x, y, z) tuples (floats for coords)
    """
    natoms = None
    box_bounds = {"x":None, "y":None, "z":None}
    atoms_section = False
    atoms = []
    # We'll record how many atoms we saw, for sanity
    num_atoms_parsed = 0

    with open(filename, 'r') as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            # 1) Look for “<number> atoms”
            m = re.match(r"^(\d+)\s+atoms\s*$", line)
            if m:
                natoms = int(m.group(1))
                continue

            # 2) Box bounds: “xlo xhi”
            m = re.match(r"^([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+xlo\s+xhi\s*$", line)
            if m:
                box_bounds["x"] = (float(m.group(1)), float(m.group(2)))
                continue
            m = re.match(r"^([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+ylo\s+yhi\s*$", line)
            if m:
                box_bounds["y"] = (float(m.group(1)), float(m.group(2)))
                continue
            m = re.match(r"^([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+([+-]?\d*\.?\d+(?:[eE][+-]?\d+)?)\s+zlo\s+zhi\s*$", line)
            if m:
                box_bounds["z"] = (float(m.group(1)), float(m.group(2)))
                continue

            # 3) Detect start of “Atoms” section
            if line.startswith("Atoms"):
                # Next non-blank line is the header of columns (we ignore it),
                # then successive lines are “id  type  x  y  z ...”
                atoms_section = True
                # Skip the next line (usually the column header)
                next(f)
                continue

            # 4) If we are in Atoms section, parse until we have read natoms lines
            if atoms_section:
                parts = line.split()
                if len(parts) < 5:
                    # If <5 columns, probably blank or end
                    atoms_section = False
                    continue
                # LAMMPS “atomic” style: id, type, x, y, z
                try:
                    at_id   = int(parts[0])
                    at_type = int(parts[2])
                    x = float(parts[3])
                    y = float(parts[4])
                    z = float(parts[5])
                except ValueError:
                    # Malformed line → stop parsing Atoms
                    atoms_section = False
                    continue
                atoms.append((at_id, at_type, x, y, z))
                num_atoms_parsed += 1
                if natoms is not None and num_atoms_parsed >= natoms:
                    # We have read all atoms → exit Atoms section
                    atoms_section = False

    if natoms is None:
        raise RuntimeError(f"Could not find number of atoms in {filename}")
    if any(box_bounds[dim] is None for dim in ("x","y","z")):
        raise RuntimeError(f"Failed to parse box bounds from {filename}")
    if len(atoms) != natoms:
        raise RuntimeError(f"Expected {natoms} atoms, but parsed {len(atoms)} in {filename}")
    return natoms, (box_bounds["x"], box_bounds["y"], box_bounds["z"]), atoms

def write_mlip_cfg(outfile, natoms, box, atoms):
    (xlo, xhi), (ylo, yhi), (zlo, zhi) = box
    # Build lattice vectors (orthorhombic):
    ax = (xhi - xlo, 0.0,       0.0)
    by = (0.0,       yhi - ylo, 0.0)
    cz = (0.0,       0.0,       zhi - zlo)

    with open(outfile, 'w') as out:
        out.write("BEGIN_CFG\n")
        out.write(" Size\n")
        out.write(f"    {natoms}\n")
        out.write(" Supercell\n")
        out.write(f"    {ax[0]:15.8f} {ax[1]:15.8f} {ax[2]:15.8f}\n")
        out.write(f"    {by[0]:15.8f} {by[1]:15.8f} {by[2]:15.8f}\n")
        out.write(f"    {cz[0]:15.8f} {cz[1]:15.8f} {cz[2]:15.8f}\n")
        out.write(" AtomData:  id  type  cartes_x  cartes_y  cartes_z\n")
        # Sort atoms by ID (optional, but often nicer)
        atoms_sorted = sorted(atoms, key=lambda x: x[0])
        for (at_id, at_type, x, y, z) in atoms_sorted:
            # Map numeric type → element symbol:
            out.write(f"    {at_id:4d}  {at_type-1:4d}   {x:10.6f}   {y:10.6f}   {z:10.6f}\n")
        out.write("END_CFG\n")
```

## Plot force energy

```python
def plot_force_energy(sf, wf, f_rmse, e_rmse):    ## Usage
    filename = f'./cal_train_E0_sf{sf}_f{wf}.cfg.0'
    E_MTP,supercell, F_MTP, S_MTP = load_multi_section_cfg_no_grade(filename)
    # Usage
    filename = './train_E_0.cfg'
    E_DFT, supercell, F_DFT, S_DFT = load_multi_section_cfg_no_grade(filename)
    F_DFT = np.array(F_DFT, dtype=object)
    F_MTP = np.array(F_MTP, dtype=object)
    x = np.concatenate([arr[:,5:].flatten() for arr in F_DFT])
    y = np.concatenate([arr[:,5:].flatten() for arr in F_MTP])
    
    fig, axs = plt.subplots(2, 2, figsize=(singlecol*2, singlecol*2), dpi=300)
    plt.subplots_adjust(wspace=0.35, hspace=0.35)
    
    # === Top-left: Force parity plot ===
    ax1 = axs[0, 0]
    from scipy import stats
    
    x = np.concatenate([arr[:,5:].flatten() for arr in F_DFT])
    y = np.concatenate([arr[:,5:].flatten() for arr in F_MTP])
    import math
    FDFT = []
    for i in range(len(E_DFT)):
        x = np.concatenate([arr[5:].flatten() for arr in F_DFT[i]])
        for j in range(len(x)):
            FDFT.append(math.sqrt(np.linalg.norm(x[3*j : 3*j+3])))
    FMTP = []
    for i in range(len(E_MTP)):
        x = np.concatenate([arr[5:].flatten() for arr in F_MTP[i]])
        for j in range(len(x)):
            FMTP.append(math.sqrt(np.linalg.norm(x[3*j : 3*j+3])))
            
    FMTP=np.array(FMTP)
    FDFT=np.array(FDFT)
    slope, intercept, r_value, p_value, std_err = stats.linregress(FDFT, FMTP)
    
    ax1.text(0.05, 0.95, f'R² = {r_value**2:.6f}', transform=ax1.transAxes)
    ax1.text(0.05, 0.90, f'RMSE = {f_rmse} (eV/Å)', transform=ax1.transAxes)
    min_val = min(FDFT.min(), FMTP.min())
    max_val = max(FDFT.max(), FMTP.max())
    ax1.plot([min_val-5, max_val+5], [min_val-5, max_val+5], 'r--')
    ax1.scatter(FDFT, FMTP, alpha=0.3)
    ax1.set_xlabel(r'$|F_{\mathrm{DFT}}|$ (eV/Å)')
    ax1.set_ylabel(r'$|F_{\mathrm{MTP}}|$ (eV/Å)')
    ax1.set_aspect('equal')
    ax1.set_xlim(-5, 155)
    ax1.set_ylim(-5, 155)
    
    # === Top-right: Energy parity plot ===
    ax2 = axs[0, 1]
    slope, intercept, r_value, p_value, std_err = stats.linregress(E_DFT, E_MTP)
    rmse = calculate_rmse(E_MTP, E_DFT)
    
    ax2.text(0.05, 0.95, f'R² = {r_value**2:.6f}', transform=ax2.transAxes)
    ax2.text(0.05, 0.90, f'RMSE = {e_rmse} (eV/atom)', transform=ax2.transAxes)
    
    min_val = min(E_DFT.min(), E_MTP.min())
    max_val = max(E_DFT.max(), E_MTP.max())
    ax2.plot([min_val-50, max_val+50], [min_val-50, max_val+50], 'r--')
    ax2.scatter(E_DFT, E_MTP, alpha=0.3)
    ax2.set_xlabel(r'$E_{\mathrm{DFT}}$ (eV)')
    ax2.set_ylabel(r'$E_{\mathrm{MTP}}$ (eV)')
    ax2.set_aspect('equal')
    ax2.set_xlim(min_val-50, max_val+50)
    ax2.set_ylim(min_val-50, max_val+50)
    
    # === Bottom-left: Force error plot ===
    ax3 = axs[1, 0]
    ax3.scatter(FDFT, np.abs(FDFT - FMTP), alpha=0.3)
    ax3.set_xlabel(r'$|F_{\mathrm{DFT}}|$ (eV/Å)')
    ax3.set_ylabel(r'$||F_{\mathrm{DFT}}| - |F_{\mathrm{MTP}}||$ (eV/Å)')
    
    # === Bottom-right: Energy error plot ===
    ax4 = axs[1, 1]
    ax4.scatter(E_DFT, np.abs(E_DFT - E_MTP), alpha=0.3)
    ax4.set_xlabel(r'$E_{\mathrm{DFT}}$ (eV)')
    ax4.set_ylabel(r'$|E_{\mathrm{DFT}} - E_{\mathrm{MTP}}|$ (eV)')
    
    plt.tight_layout()
    fig.suptitle(f"Lv12, E<0, scale by force={sf}, force weight={wf}")
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.savefig(f'force_energy_lv12_E0_sf{sf}_f{wf}.png', dpi=300, bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats
import math


def load_and_process(mtp_filename, dft_filename):
    """Load an MTP/DFT cfg pair and return flattened force/energy arrays
    needed for the parity plots and error histograms."""
    E_MTP, supercell, F_MTP, S_MTP = load_multi_section_cfg_no_grade(mtp_filename)
    E_DFT, supercell, F_DFT, S_DFT = load_multi_section_cfg_no_grade(dft_filename)

    F_DFT = np.array(F_DFT, dtype=object)
    F_MTP = np.array(F_MTP, dtype=object)

    # Per-atom force magnitudes (kept for parity with your original code,
    # though FDFT_components / FMTP_components are what get plotted below)
    FDFT, FMTP = [], []
    for i in range(len(E_DFT)):
        x = np.concatenate([arr[5:].flatten() for arr in F_DFT[i]])
        for j in range(len(x) // 3):
            FDFT.append(math.sqrt(np.linalg.norm(x[3 * j: 3 * j + 3])))
    for i in range(len(E_MTP)):
        x = np.concatenate([arr[5:].flatten() for arr in F_MTP[i]])
        for j in range(len(x) // 3):
            FMTP.append(math.sqrt(np.linalg.norm(x[3 * j: 3 * j + 3])))
    FDFT = np.array(FDFT)
    FMTP = np.array(FMTP)

    FDFT_components = np.concatenate([
        np.concatenate([arr[5:].flatten() for arr in F_DFT[i]])
        for i in range(len(F_DFT))
    ])
    FMTP_components = np.concatenate([
        np.concatenate([arr[5:].flatten() for arr in F_MTP[i]])
        for i in range(len(F_MTP))
    ])

    atom_counts = np.array([F_DFT[i].shape[0] for i in range(len(F_DFT))])
    E_DFT_per_atom = E_DFT / atom_counts
    E_MTP_per_atom = E_MTP / atom_counts

    return {
        "FDFT_components": FDFT_components,
        "FMTP_components": FMTP_components,
        "E_DFT_per_atom": E_DFT_per_atom,
        "E_MTP_per_atom": E_MTP_per_atom,
    }
```

## Modify data file

```python
# Convert LAMMPS data file from atomic to full style
with open('h2o_16_300K_2.data', 'r') as infile:
    lines = infile.readlines()

with open('h2o_16_300K_2_full.data', 'w') as outfile:
    # Copy header until atoms line
    i = 0
    while i < len(lines) and not lines[i].strip().endswith('atoms'):
        outfile.write(lines[i])
        i += 1
    
    if i < len(lines):
        outfile.write(lines[i])  # Write atoms line
        i += 1
    
    # Write atom types line if it exists
    if i < len(lines) and 'atom types' in lines[i]:
        outfile.write(lines[i])
        i += 1
    
    # Add bond and angle information
    #outfile.write("0 bonds\n")
    #outfile.write("0 angles\n")
    #outfile.write("0 dihedrals\n")
    #outfile.write("0 impropers\n\n")
    
    # Skip to Atoms section
    while i < len(lines) and not lines[i].strip().startswith('Atoms'):
        outfile.write(lines[i])
        i += 1
    
    # Update Atoms section header
    if i < len(lines):
        outfile.write("Atoms # full\n\n")
        i += 1
    
    # Convert atom format
    while i < len(lines):
        if lines[i].strip():
            parts = lines[i].split()
            if len(parts) >= 4:  # Check if it's an atom line
                try:
                    atom_id = int(parts[0])
                    atom_type = int(parts[1])
                    x, y, z = float(parts[2]), float(parts[3]), float(parts[4])
                    charge = 0
                    molecule = 1
                    # Write in full format
                    outfile.write(f"{atom_id} {molecule} {atom_type} {x} {y} {z}\n")
                except (ValueError, IndexError):
                    outfile.write(lines[i])  # Write the line as is if it doesn't parse as an atom
            else:
                outfile.write(lines[i])  # Write non-atom lines as is
        else:
            outfile.write(lines[i])  # Write empty lines as is
        i += 1
```

```python
def convert_atomic_to_full(input_file, output_file):
    with open(input_file, 'r') as f:
        lines = f.readlines()
    
    with open(output_file, 'w') as f:
        in_atoms_section = False
        current_molecule = 0
        last_atom_type = None
        
        for line in lines:
            line = line.strip()
            
            # Change atom_style declaration if present
            if "atom_style atomic" in line:
                f.write("atom_style full\n")
                continue
                
            # Check if we're entering the Atoms section
            if line == "Atoms":
                in_atoms_section = True
                f.write(line + "\n")
                continue
                
            # Check if we're leaving the Atoms section
            if in_atoms_section and (line.startswith("Bonds") or not line):
                in_atoms_section = False
                # If no Bonds section exists, you might want to add an empty one
                if line != "Bonds" and "Bonds" not in "".join(lines):
                    f.write("\nBonds\n\n")
            
            # Process lines in the Atoms section
            if in_atoms_section and line and not line.startswith("#"):
                parts = line.split()
                if len(parts) >= 4:  # Valid atom line
                    atom_id = parts[0]
                    atom_type = parts[1]
                    coords = parts[2:]
                    
                    # Assign molecule ID - change this logic based on your system
                    # Simple approach: new molecule when atom type changes
                    if atom_type != last_atom_type:
                        current_molecule += 1
                    last_atom_type = atom_type
                    
                    # Default charge (modify as needed for your system)
                    charge = "0.0"
                    
                    # Format: atom-ID molecule-ID atom-type q x y z
                    new_line = f"{atom_id} {current_molecule} {atom_type} {charge} {' '.join(coords)}"
                    f.write(new_line + "\n")
                    continue
            
            # Write the line unchanged if not modified above
            f.write(line + "\n")
```

## min_distance

```python
import numpy as np

# [Insert your load_multi_section_cfg_no_grade function here]

def calculate_min_distances(supercell_list, atom_data_list):
    """
    Calculates the minimum interatomic distance for a list of structures
    accounting for Periodic Boundary Conditions (PBC).
    """
    min_distances = []
    
    for idx, (supercell, atom_data) in enumerate(zip(supercell_list, atom_data_list)):
        # 1. Extract Cartesian Coordinates
        # IMPORTANT: In standard MLIP .cfg files, AtomData is usually formatted as:
        # [id, type, cartes_x, cartes_y, cartes_z, fx, fy, fz]
        # So Cartesian coordinates are at indices 2, 3, and 4. 
        # Adjust the slice `[:, 2:5]` if your .cfg format differs.
        coords = atom_data[:, 2:5]
        
        # If there's only 1 atom in the cell, min distance isn't well-defined in the same way
        if len(coords) <= 1:
            min_distances.append(np.inf)
            continue
            
        # 2. Calculate pairwise differences: r_ij = r_i - r_j
        # Shape becomes (N, N, 3)
        diffs = coords[:, np.newaxis, :] - coords[np.newaxis, :, :]
        
        # 3. Apply Minimum Image Convention for PBC
        # Convert to fractional coordinates using the inverse of the supercell matrix
        inv_supercell = np.linalg.inv(supercell)
        diffs_frac = diffs @ inv_supercell
        
        # Wrap fractional coordinates to [-0.5, 0.5] to find the shortest periodic path
        diffs_frac = diffs_frac - np.round(diffs_frac)
        
        # Convert back to Cartesian coordinates
        diffs_cart = diffs_frac @ supercell
        
        # 4. Calculate Euclidean distances
        distances = np.linalg.norm(diffs_cart, axis=-1)
        
        # 5. Mask the diagonal (self-distances are always 0)
        np.fill_diagonal(distances, np.inf)
        
        # Find the absolute minimum distance in this structure
        min_dist = np.min(distances)
        min_distances.append(min_dist)
        
    return np.array(min_distances)
```

```python
import random
import sys

def read_cfg_file(filename):
    """Read and parse the CFG file"""
    with open(filename, 'r') as f:
        lines = f.readlines()
    
    # Parse the file
    parsed_data = {
        'header': [],
        'atoms': [],
        'footer': []
    }
    
    current_section = 'header'
    atom_data_started = False
    
    for line in lines:
        line = line.strip()
        
        if line.startswith('AtomData:'):
            current_section = 'atoms'
            atom_data_started = True
            parsed_data['header'].append(line)
            print(f"Found AtomData section")
            continue
        
        if atom_data_started and current_section == 'atoms':
            # Check if we've reached the end of atom data
            if line.startswith('Energy') or line.startswith('PlusStress:'):
                current_section = 'footer'
                parsed_data['footer'].append(line)
                continue
            
            # Skip empty lines
            if line == '':
                continue
                
            # Parse atom line - handle the specific spacing in your format
            parts = line.split()
            if len(parts) >= 8:  # id, type, x, y, z, fx, fy, fz
                try:
                    atom_data = {
                        'id': int(parts[0]),
                        'type': int(parts[1]),
                        'x': float(parts[2]),
                        'y': float(parts[3]),
                        'z': float(parts[4]),
                        'fx': float(parts[5]),
                        'fy': float(parts[6]),
                        'fz': float(parts[7]),
                        'original_line': line
                    }
                    parsed_data['atoms'].append(atom_data)
                    if len(parsed_data['atoms']) <= 5:  # Debug first few atoms
                        print(f"Parsed atom {atom_data['id']}: type={atom_data['type']}")
                except (ValueError, IndexError) as e:
                    print(f"Failed to parse line: '{line}' - Error: {e}")
                    # If parsing fails, treat as non-atom line
                    parsed_data['footer'].append(line)
                    current_section = 'footer'
            else:
                print(f"Line with insufficient parts ({len(parts)}): '{line}'")
            continue
        
        # Add to appropriate section
        if current_section == 'header':
            parsed_data['header'].append(line)
        elif current_section == 'footer':
            parsed_data['footer'].append(line)
    
    return parsed_data

def delete_random_atoms_of_type(parsed_data, atom_type, num_to_delete):
    """Delete random atoms of specified type"""
    # Find all atoms of the specified type
    atoms_of_type = [atom for atom in parsed_data['atoms'] if atom['type'] == atom_type]
    
    print(f"Found {len(atoms_of_type)} atoms of type {atom_type}")
    
    if len(atoms_of_type) < num_to_delete:
        print(f"Warning: Only {len(atoms_of_type)} atoms of type {atom_type} found, but {num_to_delete} requested for deletion.")
        num_to_delete = len(atoms_of_type)
    
    # Randomly select atoms to delete
    atoms_to_delete = random.sample(atoms_of_type, num_to_delete)
    atoms_to_delete_ids = [atom['id'] for atom in atoms_to_delete]
    
    print(f"Deleting atoms with IDs: {atoms_to_delete_ids}")
    
    # Remove selected atoms from the list
    remaining_atoms = [atom for atom in parsed_data['atoms'] if atom['id'] not in atoms_to_delete_ids]
    
    # Renumber the remaining atoms
    for i, atom in enumerate(remaining_atoms, 1):
        atom['id'] = i
    
    parsed_data['atoms'] = remaining_atoms
    
    # Update the size in header
    for i, line in enumerate(parsed_data['header']):
        stripped_line = line.strip()
        if stripped_line.isdigit():  # This should be the size line
            parsed_data['header'][i] = str(len(remaining_atoms))
            break
    
    return len(atoms_to_delete)

def write_cfg_file(parsed_data, output_filename):
    """Write the modified CFG file"""
    with open(output_filename, 'w') as f:
        # Write header
        f.write("BEGIN_CFG\n")
        for line in parsed_data['header']:
            f.write(f" {line}\n")
        
        # Write atoms
        for atom in parsed_data['atoms']:
            f.write(f"        {atom['id']:6d}    {atom['type']}    "
                   f"{atom['x']:12.6f}   {atom['y']:12.6f}   {atom['z']:12.6f}    "
                   f"{atom['fx']:12.6f}   {atom['fy']:12.6f}   {atom['fz']:12.6f}\n")
        
        # Write footer
        for line in parsed_data['footer']:
            f.write(f" {line}\n")

def main():
    input_file = "relaxed_Na_32_O_32.cfg.0"  # Change this to your input file name
    output_file = "Na_28_O_32.cfg"  # Change this to your desired output file name
    atom_type_to_delete = 4
    num_atoms_to_delete = 4
    
    try:
        # Read the CFG file
        print(f"Reading CFG file: {input_file}")
        parsed_data = read_cfg_file(input_file)
        
        print(f"Original number of atoms: {len(parsed_data['atoms'])}")
        
        # Delete random atoms of type 4
        deleted_count = delete_random_atoms_of_type(parsed_data, atom_type_to_delete, num_atoms_to_delete)
        
        print(f"Deleted {deleted_count} atoms of type {atom_type_to_delete}")
        print(f"New number of atoms: {len(parsed_data['atoms'])}")
        
        # Write the modified file
        write_cfg_file(parsed_data, output_file)
        print(f"Modified CFG file written to: {output_file}")
        
    except FileNotFoundError:
        print(f"Error: File '{input_file}' not found!")
    except Exception as e:
        print(f"Error: {str(e)}")

if __name__ == "__main__":
    main()
```

```python
import os
import re
import numpy as np
import pandas as pd

def load_multi_section_cfg(filename):
    energies = []
    supercell_list = []
    atom_data_list = []
    stress_data_list = []
    grade_data_list = []
    
    try:
        with open(filename, 'r') as file:
            content = file.read()
            
            # Split the content into sections
            cfg_sections = content.split('BEGIN_CFG')[1:]  # Skip the first empty split
            
            for section in cfg_sections:
                lines = section.strip().split('\n')
                
                # Extract Size
                size = int(next(line.strip() for line in lines if line.strip().isdigit()))

                # Extract Supercell
                supercell_start = next(i for i, line in enumerate(lines) if line.strip().startswith('Supercell'))
                supercell = np.array([list(map(float, line.split()[:])) for line in lines[supercell_start+1:supercell_start+3+1]])
                supercell_list.append(supercell)
                
                # Extract AtomData
                atom_data_start = next(i for i, line in enumerate(lines) if line.strip().startswith('AtomData:'))
                atom_data = np.array([list(map(float, line.split()[:])) for line in lines[atom_data_start+1:atom_data_start+size+1]])
                atom_data_list.append(atom_data)
                
                # Extract Energy
                energy_line_index = next(i for i, line in enumerate(lines) if line.strip().startswith('Energy'))
                energy = float(lines[energy_line_index+1].split()[0])
                energies.append(energy)
                
                # Extract PlusStress
                stress_line = next(line for line in lines if line.strip().startswith('PlusStress:'))
                stress = np.array(list(map(float, lines[lines.index(stress_line) + 1].split())))
                stress_data_list.append(stress)
        
        return np.array(energies), np.array(supercell_list), atom_data_list, np.array(stress_data_list)
    
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except IsADirectoryError:
        print(f"Error: '{filename}' is a directory, not a file.")
    except IOError:
        print(f"Error: Could not read file '{filename}'.")
    except Exception as e:
        print(f"Error: {e}")
        return None, None, None

filename = './vasp.cfg'
E_DFT, supercell, F_DFT, S_DFT = load_multi_section_cfg(filename)
active_set = np.where(E_DFT < -3000)

E = E_DFT[active_set[0]]
S = S_DFT[active_set[0]]
F_DFT = np.array(F_DFT, dtype=object)
F = F_DFT[active_set[0]]
supercell = supercell[active_set[0]]

def write_cfg(E, supercell, F, S, output_file):
    with open(output_file, 'w') as file:
        for cfg_num in range(E.shape[0]):
            file.write("BEGIN_CFG\n")
            file.write(" Size\n")
            file.write(f"    {F[cfg_num].shape[0]}\n")
            file.write(" Supercell\n") 
            for row in supercell[cfg_num]:
                # Format each element in the row to scientific notation with 17 decimals
                formatted_row = " ".join([f"{elem}" for elem in row])
                file.write(formatted_row + "\n")
            file.write(" AtomData:  id type       cartes_x      cartes_y      cartes_z           fx          fy          fz\n")
    
            for atom in F[cfg_num]:
                atom_id = int(atom[0])
                atom_type = int(atom[1])
                x, y, z = map(float, atom[2:5])
                fx, fy, fz = map(float, atom[5:])
                file.write(f"    {atom_id}    {atom_type}    {x:.6f}    {y:.6f}    {z:.6f}    {fx:.6f}    {fy:.6f}    {fz:.6f}\n")

            file.write(" Energy\n")
            file.write(f"    {E[cfg_num]}\n")
            file.write(" PlusStress:  xx          yy          zz          yz          xz          xy\n")
            stress = " ".join([f"{elem}" for elem in S[cfg_num]])
            file.write(f"    {stress}\n")
            file.write("END_CFG\n")
            file.write("\n")

write_cfg(E, supercell, F, S, './cleaned_vasp.cfg')
```
