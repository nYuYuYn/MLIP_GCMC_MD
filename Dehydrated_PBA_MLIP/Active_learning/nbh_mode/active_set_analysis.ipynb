```python
import numpy as np
import pandas as pd
import math
import os
import sys
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

```python
train_set = pd.read_csv('train_set.txt', header=None, sep=' ', names=['atom counts','cfg counts'])
active_set = pd.read_csv('active_set.txt', header=None, sep=' ', names=['atom counts','cfg counts'])
validation_set = pd.read_csv('validation_set.txt', header=None, sep=' ', names=['atom counts','cfg counts'])
```

```python
plt.figure(figsize=(8,8))
plt.pie(train_set['cfg counts'], labels=train_set['atom counts'], 
        autopct='%1.1f%%', labeldistance=1.05, pctdistance=0.8, startangle=140)
plt.title(f"Training set {train_set['cfg counts'].sum()} cfgs", fontsize=16)
plt.axis('equal') 
plt.savefig(f'training_set_pie.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8,8))
plt.pie(active_set['cfg counts'], labels=active_set['atom counts'], 
        autopct='%1.1f%%', labeldistance=1.05, pctdistance=0.8, startangle=140)
plt.title(f"Active set {active_set['cfg counts'].sum()} cfgs", fontsize=16)
plt.axis('equal') 
plt.savefig(f'active_set_pie.png', dpi=300, bbox_inches='tight')
```

```python
plt.figure(figsize=(8,8))
plt.pie(validation_set['cfg counts'], labels=validation_set['atom counts'], 
        autopct='%1.1f%%', labeldistance=1.05, pctdistance=0.8, startangle=140)
plt.title(f"Validation set {validation_set['cfg counts'].sum()} cfgs", fontsize=16)
plt.axis('equal') 
plt.savefig(f'Validation set.png', dpi=300, bbox_inches='tight')
```

```python
colors = [
    '#ff9999', '#66b3ff', '#99ff99', '#ffcc99', 
    '#c2c2f0', '#ffb3e6', '#ff6666', '#ffb366', 
    '#66ffb3', '#b3ff66', '#b366ff', '#66b3ff', 
    '#9999ff', '#ff99ff', '#66ccff', '#ff66cc'
]

for i in range(len(col_name)):
    plt.figure(figsize=(6,6))
    plt.pie(nbh[col_name[i]], labels=na_list, autopct='%1.1f%%', colors=colors, startangle=140)
    plt.title(f"nbh mode {col_name[i]} {nbh[col_name[i]].sum()} cfgs")
    plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
    plt.savefig(f'nbh_{col_name[i]}_pie.png')#, dpi=300, bbox_inches='tight')
```
